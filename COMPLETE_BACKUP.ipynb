```python
import pandas as pd

# Airbnb data
airbnb = pd.read_csv("./data/airbnb_hw.csv")

# Look at Price before cleaning
print(airbnb["Price"].head())

# Remove commas and turn Price into numbers
airbnb["Price"] = airbnb["Price"].astype(str).str.replace(",", "", regex=False)
airbnb["Price"] = pd.to_numeric(airbnb["Price"], errors="coerce")

print(airbnb["Price"].head())
print("Missing Price values:", airbnb["Price"].isna().sum())
```

```python
# Minnesota police data
police = pd.read_csv("./data/mn_police_use_of_force.csv")

# Look at subject_injury
print(police["subject_injury"].value_counts(dropna=False))

# Clean subject_injury
police["subject_injury"] = police["subject_injury"].str.strip()

# Proportion missing
missing = police["subject_injury"].isna().mean()

print("Proportion missing:", round(missing, 4))
print("Percent missing:", round(missing * 100, 2), "%")
```

```python
# Cross-tabulate subject injury with force type
table = pd.crosstab(
    police["force_type"],
    police["subject_injury"].fillna("Missing")
)

table
```

```python
# Percent missing for each type of force
missing_by_force = (
    police.groupby("force_type")["subject_injury"]
    .apply(lambda x: x.isna().mean())
    .sort_values(ascending=False)
)

missing_by_force
```

```python
import pandas as pd

sharks = pd.read_excel("./data/GSAF5.xls")
sharks.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

sharks = pd.read_excel(r"C:\Users\leyto\A01-data-wrangling\data\GSAF5.xls")

# Remove empty columns
sharks = sharks.dropna(axis=1, how="all")
sharks.columns = sharks.columns.str.strip()

# Clean Year
sharks["Year"] = pd.to_numeric(sharks["Year"], errors="coerce")

print("Minimum year:", sharks["Year"].min())
print("Maximum year:", sharks["Year"].max())

# Keep attacks since 1940
sharks_1940 = sharks[sharks["Year"] >= 1940].copy()

year_counts = sharks_1940["Year"].value_counts().sort_index()

year_counts.plot()
plt.xlabel("Year")
plt.ylabel("Number of Attacks")
plt.title("Shark Attacks Since 1940")
plt.show()
```

I converted Year to numeric and focused on attacks from 1940 and later. The graph shows that shark attacks generally increased over time, although the number changes a lot from year to year. There is also a noticeable decrease in the most recent years.

```python
# Clean Age
sharks_1940["Age_clean"] = pd.to_numeric(
    sharks_1940["Age"].astype("string").str.extract(r"(\d+)", expand=False),
    errors="coerce"
)

# Remove unrealistic ages
sharks_1940.loc[
    (sharks_1940["Age_clean"] <= 0) |
    (sharks_1940["Age_clean"] > 100),
    "Age_clean"
] = pd.NA

# Make histogram
sharks_1940["Age_clean"].dropna().plot.hist(bins=20)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age of Shark Attack Victims")
plt.show()
```

The histogram shows that shark attack victims are mostly younger. The largest group is around the late teens and 20s, and the number of victims generally decreases as age increases.

```python
# Clean Sex
sex = sharks_1940["Sex"].astype("string").str.strip().str.upper()

# Keep known M and F values
known_sex = sex[sex.isin(["M", "F"])]

male_prop = (known_sex == "M").mean()

print("Proportion male:", round(male_prop, 4))
print("Percent male:", round(male_prop * 100, 2), "%")
```

About 85.72% of the victims with a recorded sex were male.

```python
# Clean Type
import pandas as pd

# Load data again
sharks = pd.read_excel(r"C:\Users\leyto\A01-data-wrangling\data\GSAF5.xls")
sharks.columns = sharks.columns.str.strip()

# Clean Year and keep 1940+
sharks["Year"] = pd.to_numeric(sharks["Year"], errors="coerce")
sharks_1940 = sharks[sharks["Year"] >= 1940].copy()

# Clean Type
type_clean = sharks_1940["Type"].astype("string").str.strip().str.lower()

sharks_1940["Type_clean"] = "Unknown"

sharks_1940.loc[
    type_clean.str.contains("unprovoked", na=False),
    "Type_clean"
] = "Unprovoked"

sharks_1940.loc[
    type_clean.str.contains("provoked", na=False) &
    ~type_clean.str.contains("unprovoked", na=False),
    "Type_clean"
] = "Provoked"

print(sharks_1940["Type_clean"].value_counts())

unprovoked_prop = (sharks_1940["Type_clean"] == "Unprovoked").mean()

print("Proportion unprovoked:", round(unprovoked_prop, 4))
print("Percent unprovoked:", round(unprovoked_prop * 100, 2), "%")
```
