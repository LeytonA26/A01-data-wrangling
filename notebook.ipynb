```python
import pandas as pd

# Airbnb data
airbnb = pd.read_csv("./data/airbnb_hw.csv")

# Look at Price before cleaning
print(airbnb["Price"].head())

# Remove commas and turn Price into numbers
airbnb["Price"] = airbnb["Price"].astype(str).str.replace(",", "", regex=False)
airbnb["Price"] = pd.to_numeric(airbnb["Price"], errors="coerce")

print(airbnb["Price"].head())
print("Missing Price values:", airbnb["Price"].isna().sum())
```

```python
# Minnesota police data
police = pd.read_csv("./data/mn_police_use_of_force.csv")

# Look at subject_injury
print(police["subject_injury"].value_counts(dropna=False))

# Clean subject_injury
police["subject_injury"] = police["subject_injury"].str.strip()

# Proportion missing
missing = police["subject_injury"].isna().mean()

print("Proportion missing:", round(missing, 4))
print("Percent missing:", round(missing * 100, 2), "%")
```

```python
# Cross-tabulate subject injury with force type
table = pd.crosstab(
    police["force_type"],
    police["subject_injury"].fillna("Missing")
)

table
```

```python
# Percent missing for each type of force
missing_by_force = (
    police.groupby("force_type")["subject_injury"]
    .apply(lambda x: x.isna().mean())
    .sort_values(ascending=False)
)

missing_by_force
```

```python
import pandas as pd

sharks = pd.read_excel("./data/GSAF5.xls")
sharks.head()
```

```python
# Clean Type
import pandas as pd

# Load data again
sharks = pd.read_excel("./data/GSAF5.xls")
sharks.columns = sharks.columns.str.strip()

# Clean Year and keep 1940+
sharks["Year"] = pd.to_numeric(sharks["Year"], errors="coerce")
sharks_1940 = sharks[sharks["Year"] >= 1940].copy()

# Clean Type
type_clean = sharks_1940["Type"].astype("string").str.strip().str.lower()

sharks_1940["Type_clean"] = "Unknown"

sharks_1940.loc[
    type_clean.str.contains("unprovoked", na=False),
    "Type_clean"
] = "Unprovoked"

sharks_1940.loc[
    type_clean.str.contains("provoked", na=False) &
    ~type_clean.str.contains("unprovoked", na=False),
    "Type_clean"
] = "Provoked"

print(sharks_1940["Type_clean"].value_counts())

unprovoked_prop = (sharks_1940["Type_clean"] == "Unprovoked").mean()

print("Proportion unprovoked:", round(unprovoked_prop, 4))
print("Percent unprovoked:", round(unprovoked_prop * 100, 2), "%")
```
